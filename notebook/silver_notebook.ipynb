```python
# Ler dados de Pedidos
path= '../raw_files/olist_orders_dataset.csv'
df_order = pd.read_csv(path)
print(df_order.info())
```

```python
len(df_order)
```

```python
len(df_order[df_order['order_approved_at'].isna()])
```

```python
# remover pedidos sem data de aprovação
df_order = df_order.dropna(subset=['order_approved_at'])
```

```python
len(df_order)
```

```python
timestamp_cols = [
    'order_approved_at',
    'order_purchase_timestamp',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for column in timestamp_cols:
    df_order[column] = pd.to_datetime(df_order[column])
```

```python
df_order['ingestion_at'] = pd.Timestamp.now()
```
